```python
#Pick any school in this dataset as an initial comparison point

school = 'P.S. 034 Franklin D. Roosevelt'

school_df = df[df['School Name'] == school]

school_df.head()
```

```python
# Cleaning Data

df['Mean Score'] = pd.to_numeric(
    df['Mean Score'],
    errors='coerce'
)

print("Missing Mean Scores:", df['Mean Score'].isna().sum())
```

```python
# Select the school
school = 'P.S. 034 Franklin D. Roosevelt'

school_df = df[df['School Name'] == school]

# Descriptive statistics for the school
school_stats = school_df['Mean Score'].agg([
    'count',
    'mean',
    'median',
    'min',
    'max',
    'std'
])

print(school_stats)
```

```python
# Descriptive statistics for all schools
all_schools_stats = df['Mean Score'].agg([
    'count',
    'mean',
    'median',
    'min',
    'max',
    'std'
])

print(all_schools_stats)
```

```python
comparison = pd.DataFrame({
    'P.S. 034 Franklin D. Roosevelt': school_stats,
    'All Schools': all_schools_stats
})

print(comparison)
```

```python
#Calculate the difference in average score

school_mean = school_df['Mean Score'].mean()
overall_mean = df['Mean Score'].mean()

difference = school_mean - overall_mean

print("School Mean:", school_mean)
print("All Schools Mean:", overall_mean)
print("Difference:", difference)
```

```python
comparison_means = pd.Series({
    'P.S. 034 Franklin D. Roosevelt': school_mean,
    'All Schools': overall_mean
})

comparison_means.plot(
    kind='bar',
    title='Mean Regents Score: P.S. 034 vs. All Schools',
    xlabel='Group',
    ylabel='Mean Score'
)
```

```python
# Prepare data for seaborn bar plot
plot_data = comparison.loc['mean'].reset_index()
plot_data.columns = ['Group', 'Mean Score']

sns.barplot(
    data=plot_data,
    x='Group',
    y='Mean Score'
)

plt.title('Mean Regents Score: P.S. 034 vs. All Schools')
plt.xlabel('Group')
plt.ylabel('Mean Score')
plt.xticks(rotation=15)
plt.show()
```

```python

# Create comparison groups
school = 'P.S. 034 Franklin D. Roosevelt'

df['Comparison'] = df['School Name'].apply(
    lambda x: 'P.S. 034 Franklin D. Roosevelt'
    if x == school else 'All Other Schools'
)

# Boxplot
plt.figure(figsize=(10, 6))

sns.boxplot(
    data=df,
    x='Comparison',
    y='Mean Score'
)

plt.title('Distribution of Regents Mean Scores')
plt.xlabel('School Group')
plt.ylabel('Mean Score')

plt.show()
```
